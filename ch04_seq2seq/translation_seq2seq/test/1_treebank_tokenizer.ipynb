```python
from nltk import TreebankWordTokenizer, TreebankWordDetokenizer
```

```python
tokenizer = TreebankWordTokenizer()
detokenizer = TreebankWordDetokenizer()
```

```python
# 定义一个英文句子
text = "Significant advancements in deep learning have greatly enhanced the performance of object detection algorithms interms of both accuracy and real-time processing. Object detection focuses on identifying andlocating multiple overlap-ping objects within an image, providing precise bounding boxes for each object."
```

```python
# 分词
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
# 合并恢复句子
print(detokenizer.detokenize(tokens))
```
