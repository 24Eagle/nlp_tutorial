```python
import jieba
```

```python
text = "小明毕业于北京大学计算机系"
```

## 精确模式（默认）

```python
# 返回一个生成器对象
gen = jieba.cut(text)
print(gen)
```

```python
for word in gen:
    print(word)
```

```python
# 返回一个词的列表
word_list = jieba.lcut(text)
print(word_list)
```

## 全模式,设置cut_all=True

```python
# 返回一个生成器对象
gen = jieba.cut(text,cut_all = True)
print(gen)
for word in gen:
    print(word)
```

```python
# 返回一个词的列表
word_list = jieba.lcut(text,cut_all = True)
print(word_list)
```

## 搜索引擎模式，cut_for_search

```python
# 返回一个生成器对象
gen = jieba.cut_for_search(text)
print(gen)
for word in gen:
    print(word)
```

```python
# 返回一个词的列表
word_list = jieba.lcut_for_search(text)
print(word_list)
```
